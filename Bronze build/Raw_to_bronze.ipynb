# Config Notebook Orchestration

This notebook acts as a controller that executes a series of configuration notebooks sequentially.

Because the environment runs on serverless compute, workspace discovery APIs are unavailable.  
Instead of scanning a folder, the list of notebooks to run is supplied as an input parameter.

The notebook:
1. Reads the provided paths.
2. Executes each notebook in order.
3. Stops if any execution fails.

This keeps the process deterministic, transparent, and compatible with serverless limitations.


```python
# Databricks notebook source

# Give notebook paths separated by comma
dbutils.widgets.text("notebooks", "")

paths = dbutils.widgets.get("notebooks").strip()

if not paths:
    raise ValueError("Provide notebook paths separated by comma")

notebooks = [p.strip() for p in paths.split(",") if p.strip()]

print("Will run:")
for nb in notebooks:
    print(" -", nb)

for nb in notebooks:
    print(f"\nRunning: {nb}")
    dbutils.notebook.run(nb, 0)

print("\nDone.")
```
